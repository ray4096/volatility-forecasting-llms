### Consolidating Datasets and Adding Sentiment Scores

**Reasoning**:
1. I'll rename columns in `df_stock_json` for better clarity (`symbol` to `Ticker`, `Publishdate` to `Date`, `Title` to `News_Title`, `Text` to `News_Text`).
2. Convert the `Date` column to datetime objects.
3. The `financial_sentiment_lexicon 2.csv` will be converted into a dictionary for efficient lookup.
4. I will create a function `calculate_sentiment` that takes text and the lexicon, tokenizes the text, and calculates an average sentiment score based on matching words.
5. Apply this function to the `News_Title` and `News_Text` columns of the consolidated DataFrame to get sentiment scores.

````python
import re
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab') # Add this line to download the missing resource

# --- Step 1: Prepare df_stock_json ---
# Make a copy to avoid modifying the original DataFrame if it's used elsewhere
df_consolidated = df_stock_json.copy()

# Rename columns for clarity and consistency
df_consolidated = df_consolidated.rename(columns={
    'symbol': 'Ticker',
    'Publishdate': 'Date',
    'Title': 'News_Title',
    'Text': 'News_Text'
})

# Convert 'Date' column to datetime objects
df_consolidated['Date'] = pd.to_datetime(df_consolidated['Date'])

# --- Step 2: Prepare the Sentiment Lexicon ---
# Convert the lexicon DataFrame to a dictionary for faster lookups
sentiment_lexicon_dict = dict(zip(df_lexicon['Word_or_Phrase'], df_lexicon['Sentiment_Score']))

# --- Step 3: Define Sentiment Calculation Function ---
def calculate_sentiment(text, lexicon_dict):
    if pd.isna(text) or not isinstance(text, str):
        return 0.0 # Return 0 for missing or non-string text

    # Convert text to lowercase and tokenize
    words = word_tokenize(text.lower())

    # Remove punctuation and non-alphabetic words (optional, but good for lexicon matching)
    words = [re.sub(r'[^a-z]', '', word) for word in words if word.isalpha()]

    # Get scores for words present in the lexicon
    scores = [lexicon_dict[word] for word in words if word in lexicon_dict]

    # Return the average score, or 0 if no matching words are found
    return sum(scores) / len(scores) if scores else 0.0

# --- Step 4: Apply Sentiment Scoring ---
# Calculate sentiment for News_Title
df_consolidated['Title_Sentiment'] = df_consolidated['News_Title'].apply(lambda x: calculate_sentiment(x, sentiment_lexicon_dict))

# Calculate sentiment for News_Text
df_consolidated['Text_Sentiment'] = df_consolidated['News_Text'].apply(lambda x: calculate_sentiment(x, sentiment_lexicon_dict))

# Optionally, create a combined sentiment score (e.g., average of title and text sentiment)
df_consolidated['Combined_Sentiment'] = (df_consolidated['Title_Sentiment'] + df_consolidated['Text_Sentiment']) / 2

# --- Display the first few rows of the consolidated dataset ---
print("\nFirst 5 rows of the consolidated dataset with sentiment scores:")
display(df_consolidated.head())

print("\nInfo of the consolidated dataset:")
df_consolidated.info()
````

````python
import pandas as pd
import yfinance as yf
from datetime import datetime, timedelta
import requests

# --- Re-execute preliminary steps to ensure variables are defined ---
# From cell 47bb281c: Get S&P 500 table
url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
response = requests.get(url, headers=headers)
tables = pd.read_html(response.text)
sp500_table = tables[0]

# From cell e1727944: Get top 50 S%26P 500 tickers by market cap
ticker_market_caps = []
for symbol in sp500_table['Symbol']:
    ticker = yf.Ticker(symbol)
    try:
        market_cap = ticker.info.get('marketCap', 0)
    except Exception:
        market_cap = 0
    ticker_market_caps.append({'Symbol': symbol, 'MarketCap': market_cap})
market_cap_df = pd.DataFrame(ticker_market_caps)
market_cap_df = market_cap_df.sort_values(by='MarketCap', ascending=False)
top_50_sp500_tickers_by_market_cap = market_cap_df.head(50)['Symbol'].tolist()

# From cell 79e370bc: Define date range
end_date = datetime.now()
start_date = end_date - timedelta(days=20 * 365) # Approximation for 20 years

# --- Re-populate historical_data ---
historical_data = {}
for ticker_symbol in top_50_sp500_tickers_by_market_cap:
    try:
        data = yf.download(ticker_symbol, start=start_date, end=end_date, interval='1wk', progress=False)
        if not data.empty:
            # Flatten columns if they are a MultiIndex
            if isinstance(data.columns, pd.MultiIndex):
                # This explicitly handles cases where the MultiIndex has different levels.
                # It tries to find the 'base' column names (Open, High, etc.)
                # and if not possible, it will create unique names.
                new_cols = []
                for col_name in data.columns:
                    if isinstance(col_name, tuple):
                        # Take the last element of the tuple if it's not empty, otherwise the first
                        new_col = col_name[-1] if col_name[-1] else col_name[0]
                        new_cols.append(new_col)
                    else:
                        new_cols.append(col_name)
                data.columns = new_cols

            # Standardize column names: rename 'Adj Close' to 'Close' if present
            if 'Adj Close' in data.columns:
                data = data.rename(columns={'Adj Close': 'Close'})

            # Ensure only standard columns exist and in a specific order
            required_cols = ['Date', 'Open', 'High', 'Low', 'Close', 'Volume']
            # Get 'Date' from index and rename explicitly
            data = data.reset_index()
            data = data.rename(columns={'index': 'Date'}) # yfinance uses 'Date' as index name

            # Filter and reorder columns
            # Ensure all required_cols are present, add missing ones with NaN
            for col in required_cols:
                if col not in data.columns:
                    data[col] = pd.NA

            data = data[required_cols] # Select and reorder

            data['Ticker'] = ticker_symbol
            # Drop duplicates if any, keeping the last one (for robustness)
            data = data.drop_duplicates(subset=['Date', 'Ticker'], keep='last')

            historical_data[ticker_symbol] = data
    except Exception as e:
        print(f"Error processing {ticker_symbol}: {e}")

# Concatenate all DataFrames into a single DataFrame
all_dfs_for_concat = list(historical_data.values())
if all_dfs_for_concat:
    consolidated_df = pd.concat(all_dfs_for_concat, ignore_index=True)
else:
    consolidated_df = pd.DataFrame() # Handle case where no data is downloaded

# consolidated_stock_df_final is now the consolidated_df after cleaning
consolidated_stock_df_final = consolidated_df.copy()

# Ensure 'Date' column is datetime type for merging
consolidated_stock_df_final['Date'] = pd.to_datetime(consolidated_stock_df_final['Date'])

# --- Step 2: Aggregate df_company_headlines_sentiment to weekly sentiment ---
# Ensure df_company_headlines_sentiment is defined and up-to-date. (created in cell 58fc36ba)
# Make sure 'date' column is datetime type
df_company_headlines_sentiment['date'] = pd.to_datetime(df_company_headlines_sentiment['date'])

# Function to get the week-ending date (Friday)
def get_week_ending_date(date):
    # Friday is weekday 4. pd.Timedelta handles negative days correctly for previous week
    days_until_friday = (4 - date.weekday() + 7) % 7
    return date + pd.Timedelta(days=days_until_friday)

df_company_headlines_sentiment['Week_End_Date'] = df_company_headlines_sentiment['date'].apply(get_week_ending_date)

# Aggregate sentiment to weekly mean
weekly_sentiment = df_company_headlines_sentiment.groupby(['ticker', 'Week_End_Date'])['Title_Sentiment'].mean().reset_index()
weekly_sentiment = weekly_sentiment.rename(columns={'ticker': 'Ticker', 'Week_End_Date': 'Date'})

# --- Step 3: Merge the two prepared DataFrames ---
combined_dataset = pd.merge(consolidated_stock_df_final, weekly_sentiment, on=['Date', 'Ticker'], how='left')

print("First 5 rows of the combined dataset:")
display(combined_dataset.head())

print("\nInfo of the combined dataset:")
combined_dataset.info()
````
